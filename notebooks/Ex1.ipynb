```python
#Imports
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm
from statsmodels.tsa.seasonal import seasonal_decompose, STL
from scipy.fft import fft
np.random.seed(0)  # for reproducibility

time =np.arange(144)
seasonal_12 = 21 + 20 * np.sin(2*np.pi*time/12)
trend = time * 2.65 + 100
residuals = np.random.normal(loc=0.0, scale=3, size=len(time))
time_series = residuals + seasonal_12 + trend
fig, ax = plt.subplots(4, 1, sharex=True, sharey=False, figsize=(7, 5))

ax[0].plot(trend, color='tab:red')
ax[0].set_ylabel("Trend")

ax[1].plot(seasonal_12, color='tab:orange')
ax[1].set_ylabel("Seasonality")

ax[2].set_ylabel("Residuals")
ax[2].plot(residuals, color='tab:green')

ax[3].plot(time_series, color='tab:blue')
ax[3].set_ylabel("Time Series")

fig.suptitle("Time")

fig.tight_layout()
plt.show()



```

```python
additive_decomposition = seasonal_decompose(x=time_series, model='additive', period=12)
```

```python
def seas_decomp_plots(original, decomposition):
    _, axes = plt.subplots(4, 1, sharex=True, sharey=False, figsize=(7, 5))
    axes[0].plot(original, label='Original')
    axes[0].legend(loc='upper left')
    axes[1].plot(decomposition.trend, label='Trend')
    axes[1].legend(loc='upper left')
    axes[2].plot(decomposition.seasonal, label='Seasonality')
    axes[2].legend(loc='upper left')
    axes[3].plot(decomposition.resid, label='Residuals')
    axes[3].legend(loc='upper left')
    plt.show()
    
seas_decomp_plots(time_series,additive_decomposition )
```
